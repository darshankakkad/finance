# Correlation matrix for stocks

```python
import pandas as pd
import numpy as np
import datetime as dt
import yfinance as yf
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def correlation_matrix(tickers):
    """
    Makes a correlation heat map of different stocks, based on data from daily returns
    """
    dfs = {t: yf.Ticker(t).history(period="1y") for t in tickers}
    close_df = pd.concat({t: df["Close"] for t, df in dfs.items()}, axis=1)
    daily_returns_df = close_df.iloc[:, :].pct_change() * 100
    daily_returns_df.replace(np.nan, 0, inplace = True)
    plt.figure(figsize = (10, 8))
    sns.heatmap(daily_returns_df.corr(), annot = True)
```

```python
correlation_matrix(["NVDA", "AAPL", "MSFT", "AMZN", "GOOGL", "META", "AVGO", "TSLA", "LLY", "WMT", "JPM", "V","ORCL","NFLX"])
```

```python
correlation_matrix(["JNJ","PLTR","V","GE"])
```
